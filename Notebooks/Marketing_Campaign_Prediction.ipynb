```python
retail_ml['Marital_Status'] = retail_ml['Marital_Status'].replace({
    'Alone': 'Single',
    'YOLO': 'Single',
    'Absurd': 'Single'
})
```

```python
retail_ml['Marital_Status'].value_counts()
```

```python
retail_ml['Dt_Customer'] = pd.to_datetime(
    retail_ml['Dt_Customer'],
    dayfirst=True
)
```

```python
reference_date = retail_ml['Dt_Customer'].max()

retail_ml['Customer_Tenure'] = (
    reference_date - retail_ml['Dt_Customer']
).dt.days
```

```python
retail_ml[['Dt_Customer', 'Customer_Tenure']].head()
```

```python
bank_ml = pd.get_dummies(
    bank_ml,
    columns=[
        'job',
        'marital',
        'education',
        'contact',
        'month',
        'poutcome'
    ],
    drop_first=True
)
```

```python
bank_ml['default'] = bank_ml['default'].map({'yes':1,'no':0})
bank_ml['housing'] = bank_ml['housing'].map({'yes':1,'no':0})
bank_ml['loan'] = bank_ml['loan'].map({'yes':1,'no':0})
bank_ml['y'] = bank_ml['y'].map({'yes':1,'no':0})
```

```python
retail_ml.drop('Dt_Customer', axis=1, inplace=True)
```

```python
retail_ml = pd.get_dummies(
    retail_ml,
    columns=[
        'Education',
        'Marital_Status'
    ],
    drop_first=True
)
```

```python
print("Bank Shape:", bank_ml.shape)
print("Retail Shape:", retail_ml.shape)
```

```python
bank_ml.head()
```

```python
retail_ml.head()
```

```python
X_bank = bank_ml.drop('y', axis=1)
y_bank = bank_ml['y']

print(X_bank.shape)
print(y_bank.shape)
```

```python
X_retail = retail_ml.drop('Response', axis=1)
y_retail = retail_ml['Response']

print(X_retail.shape)
print(y_retail.shape)
```

```python
# Train-Test split

from sklearn.model_selection import train_test_split

X_train_bank, X_test_bank, y_train_bank, y_test_bank = train_test_split(
    X_bank,
    y_bank,
    test_size=0.2,
    random_state=42,
    stratify=y_bank
)

X_train_retail, X_test_retail, y_train_retail, y_test_retail = train_test_split(
    X_retail,
    y_retail,
    test_size=0.2,
    random_state=42,
    stratify=y_retail
)
```

```python
print(X_train_bank.shape)
print(X_test_bank.shape)
```

```python
print(X_train_retail.shape)
print(X_test_retail.shape)
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)


X_train_bank_smote, y_train_bank_smote = smote.fit_resample(
    X_train_bank,
    y_train_bank
)


X_train_retail_smote, y_train_retail_smote = smote.fit_resample(
    X_train_retail,
    y_train_retail
)
```

```python
print("Bank Dataset")
print(y_train_bank_smote.value_counts())

print("\nRetail Dataset")
print(y_train_retail_smote.value_counts())
```

```python
# Logistic Regression

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_bank_scaled = scaler.fit_transform(X_train_bank_smote)
X_test_bank_scaled = scaler.transform(X_test_bank)

lr_bank = LogisticRegression(
    random_state=42,
    max_iter=5000
)

lr_bank.fit(X_train_bank_scaled, y_train_bank_smote)

bank_pred = lr_bank.predict(X_test_bank_scaled)
```

```python
print("BANK DATASET")
print("Accuracy :", accuracy_score(y_test_bank, bank_pred))
print("Precision:", precision_score(y_test_bank, bank_pred))
print("Recall   :", recall_score(y_test_bank, bank_pred))
print("F1 Score :", f1_score(y_test_bank, bank_pred))

print("\nConfusion Matrix")
print(confusion_matrix(y_test_bank, bank_pred))

print("\nClassification Report")
print(classification_report(y_test_bank, bank_pred))
```

```python
from sklearn.preprocessing import StandardScaler

scaler_retail = StandardScaler()

X_train_retail_scaled = scaler_retail.fit_transform(X_train_retail_smote)
X_test_retail_scaled = scaler_retail.transform(X_test_retail)

lr_retail = LogisticRegression(
    random_state=42,
    max_iter=5000
)

lr_retail.fit(X_train_retail_scaled, y_train_retail_smote)

retail_pred = lr_retail.predict(X_test_retail_scaled)

print("RETAIL DATASET")
print("Accuracy :", accuracy_score(y_test_retail, retail_pred))
print("Precision:", precision_score(y_test_retail, retail_pred))
print("Recall   :", recall_score(y_test_retail, retail_pred))
print("F1 Score :", f1_score(y_test_retail, retail_pred))

print("\nConfusion Matrix")
print(confusion_matrix(y_test_retail, retail_pred))

print("\nClassification Report")
print(classification_report(y_test_retail, retail_pred))
```
